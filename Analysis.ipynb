```python
import pandas as pd

# 1. Read the file normally (which loads it as a single-column DataFrame)
raw_df = pd.read_csv('TechTrend_Orders_2026-07-02.csv')

# 2. Extract that single massive column header string and split it cleanly by commas
# This takes 'Order ID,Date,Product ID...' and turns it into a clean Python list
giant_header_string = raw_df.columns[0]
clean_columns = [col.replace('"', '').strip() for col in giant_header_string.split(',')]

# 3. Force the data rows to split by commas as well
# We take the data inside that single column and expand it into multiple columns
df = raw_df[giant_header_string].str.split(',', expand=True)

# 4. Assign our clean list of column names to the new DataFrame
df.columns = clean_columns

# 5. Clean up string quotes from the actual data cells
for col in df.columns:
    df[col] = df[col].astype(str).str.replace('"', '').str.strip()

# 6. Convert fields to their proper data types for your calculations
df['Date'] = pd.to_datetime(df['Date'])

numeric_cols = ['Quantity Sold', 'Retail Price (ZAR)', 'Cost Price (ZAR)', 'Revenue (ZAR)', 'COGS (ZAR)', 'Gross Profit (ZAR)']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Check to ensure it is finally fixed
print("Success! Clean column structure achieved:")
print(df.columns.tolist())
print("\nFirst row verification:")
print(df.head(1))
```

# Chart 1: <span style = "color: teal;"> Customer Segment </span> vs. <span style = "color: coral;"> Product Category </span> (Revenue Matrix)

```python
import plotly.express as px

# Create the pivot matrix
matrix_df = pd.crosstab(
    df['Customer Segment'], 
    df['Category'], 
    values=df['Revenue (ZAR)'], 
    aggfunc='sum'
).round(2)

# Build the plot
fig_heatmap = px.imshow(
    matrix_df,
    labels=dict(x="Product Category", y="Customer Segment", color="Revenue (ZAR)"),
    x=matrix_df.columns,
    y=matrix_df.index,
    text_auto='.2s', 
    color_continuous_scale='Viridis',
    title='<b>Revenue Distribution Matrix by Segment & Category (ZAR)</b>'
)

fig_heatmap.update_layout(title_x=0.5)
fig_heatmap.show()
```

```python
# Aggregate columns for product profiles
prod_perf = df.groupby('Product Name').agg(
    Revenue=('Revenue (ZAR)', 'sum'),
    Gross_Profit=('Gross Profit (ZAR)', 'sum'),
    Quantity_Sold=('Quantity Sold', 'sum')
).reset_index()

prod_perf['Margin %'] = (prod_perf['Gross_Profit'] / prod_perf['Revenue']) * 100

# Plot using a Red-to-Green performance spectrum
fig_products = px.scatter(
    prod_perf,
    x='Quantity_Sold',
    y='Revenue',
    color='Margin %',
    hover_name='Product Name',
    color_continuous_scale='RdYlGn', 
    title='<b>Product Drilldown: Volume vs. Revenue vs. Profit Margin</b>',
    labels={'Quantity_Sold': 'Units Sold', 'Revenue': 'Total Revenue (ZAR)'}
)

fig_products.update_layout(title_x=0.5)
fig_products.show()
```

# Chart 4: <Span style = "color: teal;"> MoM (Month-over-Month Revenue Growth Trend) </span>

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. Group data by month and sum the revenue
monthly_data = df.set_index('Date').resample('ME')['Revenue (ZAR)'].sum().reset_index()

# 2. Format Month column to a clean string format (e.g., 'Jan 2024') for clear X-axis labeling
monthly_data['Month_Label'] = monthly_data['Date'].dt.strftime('%b %Y')

# 3. Calculate the Month-over-Month percentage growth rate
monthly_data['MoM Growth %'] = monthly_data['Revenue (ZAR)'].pct_change() * 100

# 4. Create a dual-axis chart figure layout
fig_mom = make_subplots(specs=[[{"secondary_y": True}]])

# Add Bar Chart: Total monthly revenue volume (Left Y-Axis)
fig_mom.add_trace(
    go.Bar(
        x=monthly_data['Month_Label'],
        y=monthly_data['Revenue (ZAR)'],
        name="Total Revenue (ZAR)",
        marker_color='#34495E',
        opacity=0.85,
        hovertemplate="Month: %{x}<br>Revenue: R%{y:,.2f}<extra></extra>"
    ),
    secondary_y=False,
)

# Add Line Chart: MoM Growth % trajectory (Right Y-Axis)
fig_mom.add_trace(
    go.Scatter(
        x=monthly_data['Month_Label'],
        y=monthly_data['MoM Growth %'],
        name="MoM Growth %",
        mode="lines+markers+text",
        line=dict(color='#E74C3C', width=3),
        marker=dict(size=8),
        text=monthly_data['MoM Growth %'].round(1).astype(str) + '%', # Add text labels next to points
        textposition="top center",
        hovertemplate="Month: %{x}<br>Growth Rate: %{y:.1f}%<extra></extra>"
    ),
    secondary_y=True,
)

# 5. Fine-tune axis formatting and chart appearance
fig_mom.update_layout(
    title='<b>TechTrend Monthly Performance: Revenue Volume vs. MoM Growth Rate</b>',
    title_x=0.5,
    xaxis_title='Timeline',
    legend=dict(x=0.01, y=0.99, bgcolor='rgba(255,255,255,0.6)'),
    height=600,
    width=950
)

# Set descriptive titles for both y-axes
fig_mom.update_yaxes(title_text="<b>Total Revenue (ZAR)</b>", secondary_y=False, tickformat=",.0f")
fig_mom.update_yaxes(title_text="<b>Growth Rate (%)</b>", secondary_y=True, ticksuffix="%")

# Display chart
fig_mom.show()
```

# <span style = "color: teal;"> Customer Segmentation Growth Dynamics </span> & <span style = "color: coral;"> Forecasting Baseline Trendline Across the Next 3 Months of Revenue Values </span>

Part 1: <span style = "color: teal;"> Customer Segmentation Growth Dynamics </span>

This  block calculates a monthly pivot matrix showing total sales per customer group, determines the dynamic month-over-month growth rate for each unique segment, and visualizes it as an interactive multi-line chart.

```python
import pandas as pd
import plotly.express as px

# 1. Resample and group by Month and Customer Segment
segment_monthly = df.set_index('Date').groupby([
    pd.Grouper(freq='ME'), 'Customer Segment'
])['Revenue (ZAR)'].sum().reset_index()

# 2. Add a clean timeline string label for the X-axis
segment_monthly['Month'] = segment_monthly['Date'].dt.strftime('%b %Y')

# 3. Calculate separate Month-over-Month growth percentages inside each segment group
segment_monthly['MoM Growth %'] = segment_monthly.groupby('Customer Segment')['Revenue (ZAR)'].pct_change() * 100

# 4. Plot dynamic trajectories
fig_seg_growth = px.line(
    segment_monthly.dropna(subset=['MoM Growth %']), # Drops the first month since it has no baseline
    x='Month',
    y='MoM Growth %',
    color='Customer Segment',
    markers=True,
    title='<b>Customer Segment Performance: Monthly Growth Velocity Dynamics</b>',
    labels={'MoM Growth %': 'Month-over-Month Growth Rate (%)', 'Month': 'Timeline'}
)

fig_seg_growth.update_layout(title_x=0.5, yaxis_ticksuffix='%', height=500)
fig_seg_growth.show()
```

Part 2: <span style = "color:teal;"> Automated 3-Month Forward Baseline Revenue Forecast </span>

To mathematically project a baseline trend without external dependencies, this script fits a linear trend model to your actual historical data points, generates the future 3 months of date keys, and appends the forecasted values to an interactive line visualization.

```python
import numpy as np
import plotly.graph_objects as go

# 1. Aggregate historical total revenue by month
hist_monthly = df.set_index('Date').resample('ME')['Revenue (ZAR)'].sum().reset_index()
hist_monthly['Type'] = 'Historical Revenue'

# 2. Map timestamps to an ordered sequence of integers for calculation mapping
hist_monthly['Time_Index'] = np.arange(len(hist_monthly))

# 3. Compute Linear Regression parameters (y = mx + c) using numpy polyfit
slope, intercept = np.polyfit(hist_monthly['Time_Index'], hist_monthly['Revenue (ZAR)'], 1)

# 4. Generate future chronological parameters for the next 3 months
last_hist_date = hist_monthly['Date'].max()
future_dates = pd.date_range(start=last_hist_date + pd.offsets.MonthEnd(1), periods=3, freq='ME')

future_indices = np.arange(len(hist_monthly), len(hist_monthly) + 3)
future_forecast = (slope * future_indices) + intercept

# 5. Build a clean forecast DataFrame container
forecast_df = pd.DataFrame({
    'Date': future_dates,
    'Revenue (ZAR)': future_forecast,
    'Type': '3-Month Baseline Forecast',
    'Time_Index': future_indices
})

# 6. Merge historical rows with the future projection rows
combined_forecast = pd.concat([hist_monthly, forecast_df], ignore_index=True)
combined_forecast['Month_Label'] = combined_forecast['Date'].dt.strftime('%b %Y')

# 7. Render dual-phase visualization layout
fig_forecast = px.line(
    combined_forecast,
    x='Month_Label',
    y='Revenue (ZAR)',
    color='Type',
    line_dash='Type', # Automatically applies a dashed look to distinguish the forecast segment
    markers=True,
    color_discrete_map={'Historical Revenue': '#2C3E50', '3-Month Baseline Forecast': '#E67E22'},
    title='<b>E-Commerce Revenue Trajectory & Automated 3-Month Linear Forecast</b>'
)

fig_forecast.update_layout(title_x=0.5, xaxis_title='Timeline', yaxis_title='Revenue (ZAR)', height=550)
fig_forecast.show()
```

# <span style = "color: teal;"> Region Anomaly Detection </span>

This script aggregates monthly revenue by Province, calculates the Month-over-Month (MoM) growth rates specifically for each region, and automatically filters for anomalies (defined as any month where a province experienced a sudden revenue drop of more than 15%). It then renders an interactive chart highlighting these exact trouble spots.

```python
import pandas as pd
import plotly.express as px

# 1. Group data by Month and Province to get a clean regional timeline
prov_monthly = df.set_index('Date').groupby([
    pd.Grouper(freq='ME'), 'Province'
])['Revenue (ZAR)'].sum().reset_index()

# 2. Add human-readable month labels
prov_monthly['Month_Label'] = prov_monthly['Date'].dt.strftime('%b %Y')

# 3. Calculate separate MoM growth percentages inside each individual Province
prov_monthly['MoM Growth %'] = prov_monthly.groupby('Province')['Revenue (ZAR)'].pct_change() * 100

# 4. Define an anomaly threshold (e.g., negative growth worse than -15%)
DROP_THRESHOLD = -15.0
anomalies = prov_monthly[prov_monthly['MoM Growth %'] <= DROP_THRESHOLD].copy()

# 5. Print a quick text summary of the anomalies found for easy scanning
print(f"--- REGIONAL ANOMALIES DETECTED (Drops worse than {DROP_THRESHOLD}%) ---")
if anomalies.empty:
    print("No severe regional anomalies detected in this timeframe.")
else:
    print(anomalies[['Month_Label', 'Province', 'Revenue (ZAR)', 'MoM Growth %']].to_string(index=False))

# 6. Define your custom color palette sequence (swapped duplicate red for green)
custom_colors = ['teal', 'coral', 'white', 'black', 'red', 'blue', 'orange', 'grey', 'green']

# 7. Plot all regional growth lines, highlighting where the dips occur
fig_anomaly = px.line(
    prov_monthly.dropna(subset=['MoM Growth %']),
    x='Month_Label',
    y='MoM Growth %',
    color='Province',
    color_discrete_sequence=custom_colors,
    markers=True,
    title='<b>Regional Performance: Monthly Growth Rates & Anomaly Tracking</b>',
    labels={'MoM Growth %': 'Month-over-Month Growth Rate (%)', 'Month_Label': 'Timeline'}
)

# Overlay a horizontal red reference line indicating our warning threshold level
fig_anomaly.add_hline(y=DROP_THRESHOLD, line_dash="dash", line_color="red", annotation_text="Anomaly Alert Line (-15%)")

fig_anomaly.update_layout(title_x=0.5, yaxis_ticksuffix='%', height=600, width=1000)
fig_anomaly.show()
```
